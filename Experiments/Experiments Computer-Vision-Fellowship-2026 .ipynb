# Experiment 1: Color Space Conversion
# Aim

# To convert an image into different color spaces.

```python
import cv2
import matplotlib.pyplot as plt

# Read image
image = cv2.imread("OIP.jpg")

# Convert BGR to RGB
rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Convert to Grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Convert to HSV
hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

# Display images
plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(rgb)
plt.title("RGB")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(hsv)
plt.title("HSV")
plt.axis("off")

plt.show()
##Explanation
RGB displays the normal colored image.
Grayscale converts it into black and white shades.
HSV separates color, brightness, and saturation.
```

# Experiment 2: Image Filtering
# Aim

# To remove noise using blur filters.

```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread("OIP.jpg")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

gaussian = cv2.GaussianBlur(image, (9,9), 0)

median = cv2.medianBlur(image, 5)

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(image)
plt.title("Original")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(gaussian)
plt.title("Gaussian Blur")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(median)
plt.title("Median Blur")
plt.axis("off")

plt.show()
Explanation
Gaussian Blur smooths the image.
Median Blur removes salt-and-pepper noise.
Both reduce unwanted noise before further processing.
```

# Experiment 3: Edge Detection Aim

```python

import cv2
import matplotlib.pyplot as plt

image = cv2.imread("OIP.jpg")

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

edges = cv2.Canny(gray, 100, 200)

plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(edges, cmap="gray")
plt.title("Canny Edge Detection")
plt.axis("off")

plt.show()
##Explanation
##Canny detects edges in an image.
##It highlights the boundaries of objects.
##It is widely used in computer vision applications.
```

# Experiment 4: Contour Detection
# Aim

# To detect the outlines of objects.


```python
\
import cv2
import matplotlib.pyplot as plt

image = cv2.imread("OIP.jpg")

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

contours, hierarchy = cv2.findContours(
    thresh,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

output = image.copy()

cv2.drawContours(output, contours, -1, (0,255,0), 2)

plt.imshow(cv2.cvtColor(output, cv2.COLOR_BGR2RGB))
plt.title("Contours")
plt.axis("off")
plt.show()

print("Number of Contours:", len(contours))
##Explanation
##Thresholding converts the image into black and white.
##findContours() detects the boundaries of objects.
##drawContours() draws the detected outlines.
```

# Experiment 5: Shape Detection
# Aim

# To detect basic geometric shapes.

```python

import cv2
import matplotlib.pyplot as plt

image = cv2.imread("OIP.jpg")

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

contours, _ = cv2.findContours(
    thresh,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

for contour in contours:

    approx = cv2.approxPolyDP(
        contour,
        0.02 * cv2.arcLength(contour, True),
        True
    )

    sides = len(approx)

    if sides == 3:
        shape = "Triangle"

    elif sides == 4:
        shape = "Rectangle"

    elif sides > 4:
        shape = "Circle"

    else:
        shape = "Unknown"

    x, y, w, h = cv2.boundingRect(approx)

    cv2.putText(
        image,
        shape,
        (x, y-10),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.6,
        (255,0,0),
        2
    )

    cv2.drawContours(image, [approx], -1, (0,255,0), 2)

plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Shape Detection")
plt.axis("off")
plt.show()
##Explanation
##Contours are detected from the thresholded image.
##The number of corners is used to identify the shape.
##The detected shape name is displayed on the image.
```
